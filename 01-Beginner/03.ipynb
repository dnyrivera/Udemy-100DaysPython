```python
print("Welcome to Rollercoaster")
height = int(input("What is your height in cm?: "))

if height >=120:
    total_amount = 0 
    print("You can ride the rollercoaster")
    age = int(input("What is your age?: "))
    if age <= 12: 
        total_amount = 5
        print("Child Tickets are $5.0")
    elif age <= 18:
        total_amount = 7
        print("Youth Tickets are $7.0")
    else:
        total_amount = 12
        print("Adult Tickets are $12.0")

    take_picture = input("Do you want to have a photo take? Type y for Yes and n for No.").lower()

    if take_picture == "y":
        total_amount += 3

    print(f"Your final bill is {total_amount}")

else:
    print("Sorry you have to grow taller before you can ride.")

```
